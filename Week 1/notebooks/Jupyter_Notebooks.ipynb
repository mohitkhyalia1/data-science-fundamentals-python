## 1. What is a Jupyter Notebook?

A Jupyter notebook is a document that mixes **live code**, **text**, **equations**, and **plots**
all in one place. It runs through a browser and communicates with a *kernel* (a Python process)
in the background.

The `.ipynb` file is actually just JSON under the hood — you can open it with any text editor.

**Two main cell types:**
- **Code cells** → runs Python (or other languages), shows output below
- **Markdown cells** → renders formatted text, headings, equations, images

## 3. Magic Commands

Magic commands start with `%` (line magic) or `%%` (cell magic). They're not Python —
they're instructions to the Jupyter kernel itself.

````python
# List all available magic commands
%lsmagic
````

### Timing Code — `%time` and `%timeit`

`%timeit` runs the code multiple times and gives you a reliable average. Super useful
for comparing approaches — e.g., list vs NumPy array operations.

````python
# %time — run once and report
%time sum(range(1_000_000))
````

````python
# %timeit — run many times for a reliable measurement
%timeit sum(range(10_000))
````

````python
# Compare Python list sum vs NumPy
import numpy as np
arr = np.arange(10_000)
lst = list(range(10_000))

print('Python list sum:')
%timeit sum(lst)

print('NumPy array sum:')
%timeit arr.sum()
````

## 5. Markdown in Jupyter

Markdown makes notebooks readable as actual documents, not just code files.
Here's a quick reference for the syntax I use most.

### Markdown Quick Reference

**Headers:** `# H1` `## H2` `### H3`

**Bold:** `**text**` → **text**

**Italic:** `*text*` → *text*

**Code inline:** `` `code` `` → `code`

**Math (LaTeX):** `$E = mc^2$` → $E = mc^2$

**Block math:**
```
$$\sum_{i=1}^{n} x_i = \bar{x} \cdot n$$
```
→ $$\sum_{i=1}^{n} x_i = \bar{x} \cdot n$$

**Table:**
```
| Col1 | Col2 |
|------|------|
| A    | 1    |
```

**Blockquote:** `> text`

**Horizontal rule:** `---`

## 6. Displaying Rich Output

Jupyter can display HTML, images, videos, and more using the `IPython.display` module.

````python
from IPython.display import display, HTML, Markdown

# Display formatted HTML
display(HTML('<h3 style="color: steelblue;">Hello from HTML!</h3>'))

# Render Markdown programmatically
display(Markdown('**Bold text** and *italic* rendered from Python'))

# The last expression in a cell auto-displays
{'name': 'Aryan', 'year': 1}
````

## 7. Useful Jupyter Tips

Things I wish I had known earlier:

1. **Restart & Run All** — always do this before submitting a notebook.
   It ensures your notebook runs top to bottom without hidden state.

2. **Variable inspector** — if you install the `jupyter_contrib_nbextensions` package,
   you get a Variable Inspector that shows all current variables and their types.

3. **nbconvert** — you can export notebooks to PDF, HTML, or Python scripts:
   ```bash
   jupyter nbconvert --to html my_notebook.ipynb
   jupyter nbconvert --to script my_notebook.ipynb
   ```

4. **Collapse output** — click the blue bar to the left of long outputs to collapse them.

5. **Google Colab extras** — Colab has its own magic for mounting Google Drive:
   ```python
   from google.colab import drive
   drive.mount('/content/drive')
   ```

## 8. Notebook Best Practices

Good notebooks tell a story. Here's what I try to follow:

- Start with a title cell explaining what the notebook covers
- Keep cells focused — one idea per cell
- Use Markdown cells to explain *why*, not just *what*
- Don't leave dead cells with errors
- Restart kernel and run all before committing to git
- Number your sections so it's easy to navigate
- Add a 'References' or 'Further Reading' section at the end
