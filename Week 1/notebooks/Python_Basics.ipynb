## 1. Variables and Data Types

Python is dynamically typed — you don't declare a type, Python figures it out.
What I found interesting: everything in Python is an *object*, even integers.

```python
# Basic types
x = 42          # int
pi = 3.14159    # float
name = 'Aryan'  # str
flag = True     # bool

print(type(x), type(pi), type(name), type(flag))

# Python's id() shows memory address — same small ints are cached
a = 5
b = 5
print('Same object?', a is b)  # True for small ints (-5 to 256)

# Type conversion
print(int('100') + 1)    # 101
print(float(7) / 2)     # 3.5
print(str(3.14))         # '3.14'
```

## 2. Strings

Strings are immutable sequences. A lot of Python's elegance comes from how naturally
you can manipulate strings using slicing and built-in methods.

```python
s = 'Hello, IIT Bombay!'

# Slicing
print(s[0:5])       # Hello
print(s[-7:])       # Bombay!
print(s[::-1])      # reverse

# Useful methods
print(s.upper())
print(s.replace('IIT Bombay', 'the world'))
print(s.split(', '))  # split into list

# f-strings (my favourite way to format)
score = 95.6
print(f'My score: {score:.1f}%')
```

## 3. Lists, Tuples, Sets, Dictionaries

The four main collection types. The key distinction I had to really understand:
- **Lists** → ordered, mutable
- **Tuples** → ordered, immutable (use for things that shouldn't change)
- **Sets** → unordered, no duplicates (great for membership tests)
- **Dicts** → key-value pairs, ordered since Python 3.7

```python
# Lists
marks = [85, 92, 78, 95, 88]
marks.append(91)
marks.sort(reverse=True)
print('Sorted marks:', marks)
print('Top 3:', marks[:3])

# List comprehension — much cleaner than a for loop
squared = [x**2 for x in range(1, 6)]
print('Squares:', squared)

# Dictionary
student = {'name': 'Aryan', 'branch': 'CSE', 'year': 1}
student['cgpa'] = 9.2  # add new key
print(student)

# Dict comprehension
squares_dict = {n: n**2 for n in range(1, 6)}
print(squares_dict)

# Sets
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print('Union:', a | b)
print('Intersection:', a & b)
```

## 4. Control Flow

Nothing too surprising here, but Python's `for ... in` loop is much more natural
than C-style for loops. `enumerate()` and `zip()` are genuinely useful.

```python
# if-elif-else
cgpa = 9.1
if cgpa >= 9.5:
    grade = 'Outstanding'
elif cgpa >= 8.5:
    grade = 'Excellent'
elif cgpa >= 7.5:
    grade = 'Good'
else:
    grade = 'Satisfactory'
print(f'CGPA {cgpa} → {grade}')

# enumerate — gives index AND value
subjects = ['Maths', 'Physics', 'Chemistry', 'CS']
for i, sub in enumerate(subjects, start=1):
    print(f'{i}. {sub}')

# zip — iterate over two lists together
scores = [88, 76, 91, 95]
for sub, sc in zip(subjects, scores):
    print(f'{sub}: {sc}')
```

## 5. Functions

Python functions are first-class objects — you can pass them around like variables.
I found default arguments and `*args`/`**kwargs` really useful once I understood them.

```python
# Basic function with default argument
def greet(name, greeting='Hello'):
    return f'{greeting}, {name}!'

print(greet('Aryan'))
print(greet('Prof. Kumar', greeting='Good morning'))

# *args and **kwargs
def summarize(*args, **kwargs):
    print('Positional:', args)
    print('Keyword:', kwargs)

summarize(1, 2, 3, name='Aryan', year=1)

# Lambda (anonymous function)
square = lambda x: x ** 2
print(list(map(square, range(1, 6))))

# Functions as arguments
data = [3, -1, 4, -1, 5, -9, 2, -6]
print('Sorted by abs value:', sorted(data, key=abs))
```

## 6. Recursion

A function that calls itself. The trick is always having a base case.

```python
def factorial(n):
    if n == 0 or n == 1:  # base case
        return 1
    return n * factorial(n - 1)

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print([factorial(i) for i in range(7)])
print([fibonacci(i) for i in range(10)])
```

## 7. Object-Oriented Programming

OOP in Python clicked for me when I stopped thinking of classes as blueprints
and started thinking of them as *factories for objects*. `self` is just the
instance being operated on.

```python
class Student:
    institute = 'IIT Bombay'  # class variable (shared)

    def __init__(self, name, branch, year):
        self.name = name        # instance variables
        self.branch = branch
        self.year = year
        self.courses = []

    def enroll(self, course):
        self.courses.append(course)
        print(f'{self.name} enrolled in {course}')

    def __repr__(self):
        return f'Student({self.name}, {self.branch}, Y{self.year})'


class ResearchStudent(Student):   # Inheritance
    def __init__(self, name, branch, advisor):
        super().__init__(name, branch, year=5)
        self.advisor = advisor

    def __repr__(self):
        return f'ResearchStudent({self.name}, advisor={self.advisor})'


s1 = Student('Aryan', 'CSE', 1)
s1.enroll('Data Science')
s1.enroll('Linear Algebra')
print(s1)
print('Institute:', Student.institute)

r1 = ResearchStudent('Priya', 'EE', 'Prof. Sharma')
print(r1)
print('Is Student?', isinstance(r1, Student))
```

## 8. File Handling

Always use `with open()` — it automatically closes the file even if an error occurs.

```python
# Writing to a file
lines = ['NumPy\n', 'Pandas\n', 'Matplotlib\n', 'Seaborn\n']
with open('topics.txt', 'w') as f:
    f.writelines(lines)

# Reading back
with open('topics.txt', 'r') as f:
    content = f.read()
print(content)

# Reading line by line
with open('topics.txt', 'r') as f:
    for i, line in enumerate(f, 1):
        print(f'{i}. {line.strip()}')
```

## 9. Error Handling

Rather than letting programs crash, we catch specific exceptions.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print('Error: Cannot divide by zero')
        return None
    except TypeError as e:
        print(f'Type error: {e}')
        return None
    else:
        return result
    finally:
        print('safe_divide() called')  # always runs

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide('a', 2))
```

## 10. Practice Exercises

These are problems I solved while going through the module. Try them yourself before looking at the solutions!

**Exercise 1:** Write a function that takes a list of numbers and returns
a dictionary with keys `'mean'`, `'min'`, `'max'`, and `'range'`.

**Exercise 2:** Write a function `is_palindrome(s)` that returns True if a string
reads the same forwards and backwards (ignoring case and spaces).

**Exercise 3:** Given a list of words, return a dictionary mapping each word
to its frequency count.

```python
# Exercise 1
def describe(numbers):
    return {
        'mean': sum(numbers) / len(numbers),
        'min': min(numbers),
        'max': max(numbers),
        'range': max(numbers) - min(numbers)
    }

print(describe([4, 8, 15, 16, 23, 42]))

# Exercise 2
def is_palindrome(s):
    cleaned = s.lower().replace(' ', '')
    return cleaned == cleaned[::-1]

print(is_palindrome('racecar'))        # True
print(is_palindrome('A man a plan a canal Panama'))  # True
print(is_palindrome('hello'))          # False

# Exercise 3
def word_frequency(words):
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    return freq

sentence = 'to be or not to be that is the question'.split()
print(word_frequency(sentence))
```
