## 1. The Two Core Data Structures

Pandas has two main types:
- **Series** — a 1D labelled array (like a single column)
- **DataFrame** — a 2D table of Series (like a spreadsheet)

Everything in Pandas is built around these two.

```python
import pandas as pd
import numpy as np

print('Pandas version:', pd.__version__)

# Series — 1D with index labels
cgpa = pd.Series([9.2, 8.7, 7.5, 9.6, 8.1],
                  index=['Aryan', 'Priya', 'Rahul', 'Sneha', 'Vikram'])
print('\nCGPA Series:')
print(cgpa)
print('\nMean CGPA:', cgpa.mean())
print('Top scorer:', cgpa.idxmax())
```

## 2. Creating DataFrames

You can build a DataFrame from a dictionary, a list of dicts, a NumPy array,
or by reading from a file. The dictionary method is the most common when creating
data by hand.

```python
import pandas as pd

# From a dictionary — keys become column names
data = {
    'Name'  : ['Aryan', 'Priya', 'Rahul', 'Sneha', 'Vikram', 'Meera'],
    'Branch': ['CSE', 'EE', 'ME', 'CSE', 'EE', 'ME'],
    'Year'  : [1, 2, 1, 3, 2, 1],
    'CGPA'  : [9.2, 8.7, 7.5, 9.6, 8.1, 8.9],
    'Placed': [False, True, False, True, True, False]
}

df = pd.DataFrame(data)
print(df)
print('\nShape:', df.shape)
print('Dtypes:\n', df.dtypes)
```

## 4. Selecting Data — `loc` and `iloc`

This confused me at first. The rule:
- **`loc`** — label-based indexing (uses row/column *names*)
- **`iloc`** — integer position-based (uses integer *positions*, like lists)

```python
import pandas as pd

df = pd.DataFrame({
    'Name'  : ['Aryan', 'Priya', 'Rahul', 'Sneha', 'Vikram', 'Meera'],
    'Branch': ['CSE', 'EE', 'ME', 'CSE', 'EE', 'ME'],
    'Year'  : [1, 2, 1, 3, 2, 1],
    'CGPA'  : [9.2, 8.7, 7.5, 9.6, 8.1, 8.9],
})

# Select a column
print('CGPA column:\n', df['CGPA'].values)

# Select multiple columns
print('\nName and CGPA:\n', df[['Name', 'CGPA']])

# loc — by label
print('\nRow 2 (loc):\n', df.loc[2])
print('\nRows 1-3, Name+CGPA:\n', df.loc[1:3, ['Name', 'CGPA']])

# iloc — by position
print('\nFirst 2 rows, first 3 cols (iloc):\n', df.iloc[:2, :3])
```

## 5. Filtering Rows (Boolean Indexing)

Just like NumPy — pass a boolean condition to filter rows. This is how you
query data without SQL.

```python
import pandas as pd

df = pd.DataFrame({
    'Name'  : ['Aryan', 'Priya', 'Rahul', 'Sneha', 'Vikram', 'Meera'],
    'Branch': ['CSE', 'EE', 'ME', 'CSE', 'EE', 'ME'],
    'Year'  : [1, 2, 1, 3, 2, 1],
    'CGPA'  : [9.2, 8.7, 7.5, 9.6, 8.1, 8.9],
    'Placed': [False, True, False, True, True, False]
})

# Single condition
print('CGPA >= 9.0:')
print(df[df['CGPA'] >= 9.0][['Name', 'CGPA']])

# Multiple conditions (& and |)
print('\nCSE students with CGPA > 8:')
print(df[(df['Branch'] == 'CSE') & (df['CGPA'] > 8)][['Name', 'Branch', 'CGPA']])

# isin — match against a list of values
print('\nCSE or EE students:')
print(df[df['Branch'].isin(['CSE', 'EE'])][['Name', 'Branch']])

# query() — SQL-like syntax (more readable for complex filters)
print('\nUsing query():')
print(df.query('CGPA > 8.5 and Year == 1')[['Name', 'CGPA', 'Year']])
```

## 6. Adding, Modifying, and Dropping Columns

```python
import pandas as pd

df = pd.DataFrame({
    'Name'  : ['Aryan', 'Priya', 'Rahul', 'Sneha', 'Vikram', 'Meera'],
    'CGPA'  : [9.2, 8.7, 7.5, 9.6, 8.1, 8.9],
    'Year'  : [1, 2, 1, 3, 2, 1],
})

# Add a new column
df['Grade'] = df['CGPA'].apply(lambda x: 'A+' if x >= 9 else ('A' if x >= 8 else 'B'))

# Modify a column
df['Year_Label'] = 'Year ' + df['Year'].astype(str)

# Drop a column
df = df.drop(columns=['Year'])

print(df)
```

## 7. Handling Missing Data

Real-world datasets almost always have missing values (`NaN`). This is one of
the most important skills in data cleaning.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Aryan', 'Priya', 'Rahul', 'Sneha', 'Vikram'],
    'Age' : [19, np.nan, 20, np.nan, 21],
    'CGPA': [9.2, 8.7, np.nan, 9.6, np.nan],
    'City': ['Mumbai', 'Delhi', None, 'Pune', 'Mumbai']
})

print('Original:\n', df)
print('\nNull counts:\n', df.isnull().sum())
print('\nRows with any null:\n', df[df.isnull().any(axis=1)])

# Fill with a specific value
df_filled = df.copy()
df_filled['Age'].fillna(df['Age'].mean(), inplace=True)
df_filled['CGPA'].fillna(df['CGPA'].median(), inplace=True)
df_filled['City'].fillna('Unknown', inplace=True)

print('\nAfter filling:\n', df_filled)

# Drop rows with any null
df_dropped = df.dropna()
print('\nAfter dropping nulls:\n', df_dropped)
```

## 8. GroupBy — Split, Apply, Combine

GroupBy is the pandas equivalent of SQL's `GROUP BY`. It lets you compute
summary statistics for different subgroups in your data.

```python
import pandas as pd

df = pd.DataFrame({
    'Name'  : ['Aryan','Priya','Rahul','Sneha','Vikram','Meera','Karan','Anita'],
    'Branch': ['CSE','EE','ME','CSE','EE','ME','CSE','EE'],
    'Year'  : [1, 2, 1, 3, 2, 1, 2, 3],
    'CGPA'  : [9.2, 8.7, 7.5, 9.6, 8.1, 8.9, 8.4, 9.0],
})

# Average CGPA per branch
print('Avg CGPA by branch:')
print(df.groupby('Branch')['CGPA'].mean().round(2))

# Multiple aggregations at once
print('\nMultiple aggs by branch:')
print(df.groupby('Branch')['CGPA'].agg(['mean', 'min', 'max', 'count']).round(2))

# Group by multiple columns
print('\nCount by Branch and Year:')
print(df.groupby(['Branch', 'Year'])['Name'].count().unstack(fill_value=0))
```

## 9. Sorting and Ranking

```python
import pandas as pd

df = pd.DataFrame({
    'Name'  : ['Aryan','Priya','Rahul','Sneha','Vikram','Meera'],
    'Branch': ['CSE','EE','ME','CSE','EE','ME'],
    'CGPA'  : [9.2, 8.7, 7.5, 9.6, 8.1, 8.9],
})

# Sort by one column
print('Sorted by CGPA (descending):')
print(df.sort_values('CGPA', ascending=False).reset_index(drop=True))

# Sort by multiple columns
print('\nSorted by Branch then CGPA:')
print(df.sort_values(['Branch', 'CGPA'], ascending=[True, False]).reset_index(drop=True))

# Add a rank column
df['Rank'] = df['CGPA'].rank(ascending=False).astype(int)
print('\nWith ranks:')
print(df.sort_values('Rank'))
```

## 10. Merging and Joining DataFrames

Merging is like SQL joins — you combine two DataFrames on a common key.
Pandas supports inner, left, right, and outer joins.

```python
import pandas as pd

students = pd.DataFrame({
    'ID'    : [101, 102, 103, 104],
    'Name'  : ['Aryan', 'Priya', 'Rahul', 'Sneha'],
    'Branch': ['CSE', 'EE', 'ME', 'CSE'],
})

results = pd.DataFrame({
    'ID'  : [101, 102, 103, 105],
    'CGPA': [9.2, 8.7, 7.5, 9.0],
})

# Inner join — only matching rows
print('Inner join:')
print(pd.merge(students, results, on='ID', how='inner'))

# Left join — all students, NaN if no result
print('\nLeft join:')
print(pd.merge(students, results, on='ID', how='left'))
```

## 12. Practice Exercises

**Exercise 1:** Create a DataFrame of 10 students with columns: Name, Branch,
Semester1_Marks, Semester2_Marks. Add a column 'Average' and a column 'Result'
('Pass' if average >= 40, else 'Fail').

**Exercise 2:** From that DataFrame, find:
- The branch with the highest average marks
- Students who improved (Sem2 > Sem1)

**Exercise 3:** Introduce 3 NaN values into the marks columns and handle them
by filling with the column mean.

```python
import pandas as pd
import numpy as np

np.random.seed(7)
df = pd.DataFrame({
    'Name'  : [f'Student_{i}' for i in range(1, 11)],
    'Branch': np.random.choice(['CSE', 'EE', 'ME'], 10),
    'Sem1'  : np.random.randint(30, 95, 10),
    'Sem2'  : np.random.randint(30, 95, 10),
})

df['Average'] = (df['Sem1'] + df['Sem2']) / 2
df['Result']  = df['Average'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')
df['Improved'] = df['Sem2'] > df['Sem1']

print(df.to_string(index=False))

print('\nBranch with highest avg marks:')
print(df.groupby('Branch')['Average'].mean().idxmax())

print('\nStudents who improved:')
print(df[df['Improved']]['Name'].values)
```
